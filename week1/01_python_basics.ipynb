## 2. 리스트 — 여러 값을 한 묶음으로

머신러닝 데이터는 보통 **여러 개의 값**입니다. 학생 한 명이 아니라 30명의 점수, 이미지 한 장이 아니라 수천 장. 이런 묶음을 다루는 기본 도구가 **리스트(list)** 입니다.

```python
scores = [85, 90, 78, 92, 60, 88]

print("전체:", scores)
print("첫 번째 점수:", scores[0])   # 인덱스는 0부터 시작
print("마지막 점수:", scores[-1])   # 음수는 뒤에서부터
print("개수:", len(scores))
print("합계:", sum(scores))
print("평균:", sum(scores) / len(scores))
print("최댓값:", max(scores), " 최솟값:", min(scores))
```

## 3. 조건문 — 상황에 따라 다르게

`if / elif / else` 로 조건에 따라 다른 동작을 합니다. 예를 들어 점수를 등급으로 바꾸는 것은, 나중에 모델의 예측 결과를 "합격/불합격"으로 분류하는 것과 같은 구조예요.

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"점수 {score} → 등급 {grade}")
```

## 4. 반복문 — 데이터 전체를 훑기

리스트 안의 값을 하나씩 꺼내 처리할 때 `for` 반복문을 씁니다. "모든 데이터에 같은 처리를 적용" 하는 것이 데이터 전처리의 핵심이에요.

```python
scores = [85, 90, 78, 92, 60, 88]

# 합격(70점 이상)한 사람 수 세기
count = 0
for s in scores:
    if s >= 70:
        count += 1

print("합격자 수:", count)
```

### ⭐ 핵심 예제 — 반복문으로 정규화(스케일링)

**정규화(normalization)** 는 값들을 0~1 범위로 바꾸는 것입니다. 머신러닝에서 아주 자주 하는 전처리예요.
공식은 간단합니다: `(값 - 최솟값) / (최댓값 - 최솟값)`

```python
scores = [85, 90, 78, 92, 60, 88]

lo = min(scores)
hi = max(scores)

normalized = []
for s in scores:
    n = (s - lo) / (hi - lo)
    normalized.append(round(n, 3))

print("원본:  ", scores)
print("정규화:", normalized)
# 최솟값은 0.0, 최댓값은 1.0 이 됩니다
```

## 5. 함수 — 반복되는 작업을 이름 붙여 재사용

같은 계산을 여러 번 쓸 때 **함수(function)** 로 만들면 깔끔합니다. scikit-learn 의 `fit()`, `predict()` 같은 것도 전부 함수예요.

```python
def average(numbers):
    """숫자 리스트의 평균을 반환"""
    return sum(numbers) / len(numbers)

# 재사용
print(average([80, 90, 100]))
print(average([70, 60]))
```
