## 0. 왜 NumPy인가?

머신러닝의 데이터는 사실상 전부 **배열(행렬)** 입니다. 파이썬 기본 리스트로도 계산은 되지만, 데이터가 커지면 느리고 코드도 길어져요. NumPy는 배열 전체를 한 번에, 빠르게 계산합니다.

```python
import numpy as np   # 관례적으로 np 라는 별명으로 불러옵니다

# 리스트를 배열로
scores = np.array([85, 90, 78, 92, 60, 88])
print(scores)
print("타입:", type(scores))
print("모양(shape):", scores.shape)   # (6,) → 원소 6개짜리 1차원
```

## 1. 배열 생성

자주 쓰는 생성 함수들입니다.

```python
print(np.zeros(5))          # 0으로 채운 배열
print(np.ones(3))           # 1로 채운 배열
print(np.arange(0, 10, 2))  # 0부터 10 전까지 2씩
print(np.linspace(0, 1, 5)) # 0~1을 5등분
```

---
## 2. 인덱싱 · 슬라이싱

리스트와 비슷하지만 더 강력합니다. 특히 **조건으로 골라내기(불리언 인덱싱)** 가 ML 전처리에서 매우 자주 쓰여요.

```python
scores = np.array([85, 90, 78, 92, 60, 88])

print("첫 원소:", scores[0])
print("앞 3개:", scores[:3])
print("70점 이상만:", scores[scores >= 70])   # 불리언 인덱싱
print("70점 이상 개수:", np.sum(scores >= 70)) # True를 1로 세어줌
```

## 3. 벡터 연산 & 브로드캐스팅

배열끼리, 또는 배열과 숫자를 연산하면 **모든 원소에 한 번에** 적용됩니다. 반복문이 필요 없어요. 이것이 NumPy의 핵심입니다.

```python
scores = np.array([85, 90, 78, 92, 60, 88])

print("모두 +5점:", scores + 5)        # 브로드캐스팅
print("100점 만점 대비 비율:", scores / 100)
print("평균:", scores.mean())
print("표준편차:", scores.std())
print("최대/최소:", scores.max(), scores.min())
```

### ⭐ 핵심 — 정규화를 한 줄로

1회차에서 반복문으로 짰던 정규화 `(값-최솟값)/(최댓값-최솟값)` 를, NumPy에서는 배열 전체에 한 번에 적용합니다.

```python
scores = np.array([85, 90, 78, 92, 60, 88])

normalized = (scores - scores.min()) / (scores.max() - scores.min())

print("원본:  ", scores)
print("정규화:", np.round(normalized, 3))
# 1회차 반복문 결과와 똑같습니다 — 코드는 한 줄!
```

## 4. 2차원 배열 (행렬)

ML 데이터는 보통 **행=샘플, 열=특징** 인 2차원 표입니다. 예: 학생 3명 × 과목 2개.

```python
# 행: 학생, 열: [국어, 영어]
grades = np.array([
    [80, 90],
    [70, 60],
    [95, 85],
])
print("모양:", grades.shape)      # (3, 2)
print("과목별 평균(열 기준):", grades.mean(axis=0))
print("학생별 평균(행 기준):", grades.mean(axis=1))
```

> 💡 `axis=0` 은 "행을 따라 아래로" = 열별 계산, `axis=1` 은 "열을 따라 옆으로" = 행별 계산. 처음엔 헷갈리니 위 예제를 여러 번 돌려보세요.
